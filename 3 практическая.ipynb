Подключение библиотек и загрузка данных

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

Загрузка данных

```python
df = pd.read_csv('vgsales.csv')
```

```python
df.head(10)
```

```python
print("Количество строк:", df.shape[0])
print("Количество столбцов:", df.shape[1])
print("\nНазвания столбцов:")
print(df.columns.tolist())
print("\nТипы данных:")
print(df.dtypes)
print("\nОсновные статистические характеристики:")
df.describe()
```

Поиск пропущенных значений

```python
missing = df.isnull().sum()
print(missing[missing > 0])
```

Строки, где отсутствует Yea

```python
df[df['Year'].isnull()]
```

удаление строк с пропущенным Year

```python
df_clean = df.dropna(subset=["Year"])

print("Размер исходного датасета:", df.shape)
print("Размер очищенного датасета:", df_clean.shape)
print("\nПропуски в df_clean:")
print(df_clean.isnull().sum())
```

Изучение категориальных признаков

```python
print("Жанры:", df_clean['Genre'].unique())
print("Количество жанров:", df_clean['Genre'].nunique())
```

```python
print("\nПлатформы:", df_clean['Platform'].unique())
print("Количество платформ:", df_clean['Platform'].nunique())
```

```python
print("\nИздатели (первые 20):", df_clean['Publisher'].unique()[:20])
print("Количество издателей:", df_clean['Publisher'].nunique())
```

Анализ жанров

```python
genre_count = df_clean.groupby("Genre")["Name"].count().sort_values(ascending=False)
print(genre_count)
```

Продажи по жанрам

```python
genre_sales = df_clean.groupby("Genre")["Global_Sales"].sum().sort_values(ascending=False)
print(genre_sales)
```

 Самые продаваемые игры

```python
top10_games = df_clean.sort_values("Global_Sales", ascending=False).head(10)
top10_games[["Rank", "Name", "Platform", "Genre", "Global_Sales"]]
```

Анализ платформ

Количество игр по платформам

```python
platform_count = df_clean.groupby("Platform")["Name"].count().sort_values(ascending=False)
print("Топ-10 платформ по количеству игр:")
print(platform_count.head(10))
```

Суммарные продажи по платформам

```python
platform_sales = df_clean.groupby("Platform")["Global_Sales"].sum().sort_values(ascending=False)
print("\nТоп-10 платформ по продажам:")
print(platform_sales.head(10))
```

Визуализация продаж по жанрам

```python
genre_sales_sorted = genre_sales.sort_values(ascending=True)

plt.figure(figsize=(10, 6))
plt.barh(genre_sales_sorted.index, genre_sales_sorted.values, color='steelblue')
plt.xlabel("Суммарные мировые продажи (млн)")
plt.ylabel("Жанр")
plt.title("Продажи видеоигр по жанрам")
plt.tight_layout()
plt.show()
```

Корреляция продаж

```python
corr_na_global = df_clean[["NA_Sales", "Global_Sales"]].corr()
print(corr_na_global)
```

```python
r = df_clean["NA_Sales"].corr(df_clean["Global_Sales"])
print("Коэффициент корреляции:", r)
```

Диаграмма рассеяния

```python
plt.figure(figsize=(8, 5))
plt.scatter(df_clean["NA_Sales"], df_clean["Global_Sales"], alpha=0.5, color='darkorange')
plt.xlabel("Продажи в Северной Америке, млн")
plt.ylabel("Мировые продажи, млн")
plt.title("Связь продаж в Северной Америке и мировых продаж")
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

Корреляция между регионами

```python
regions = ["NA_Sales", "EU_Sales", "JP_Sales", "Other_Sales"]
corr_regions = df_clean[regions].corr()
print(corr_regions)
```

Визуализация тепловой карты

```python
plt.figure(figsize=(6, 5))
plt.imshow(corr_regions, cmap='coolwarm', interpolation='none')
plt.colorbar()
plt.xticks(range(len(regions)), regions, rotation=45)
plt.yticks(range(len(regions)), regions)
plt.title("Корреляция продаж между регионами")
plt.tight_layout()
plt.show()
```
